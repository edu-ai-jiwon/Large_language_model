```python
def pre_df(df):
    pre_df=pd.read_csv(df)
    pre_df=pre_df[['Summary','Text']]
    pre_df.columns['title','content']

    pre_df = pre_df.dropna(subset=['content']).reset_index(drop=True)
    return pre_df
```

```python
from openai import OpenAI

client=OpenAI()
text='pre_df'
#text='안녕하세요'

response=client.embeddings.create(
    model='text-embedding-3-small',
    input=[text]
)
```

```python
def text_to_embedding(texts, model='text-embedding-3-small'):
    texts=[text.replace('\n',' ') for text in text]
    response=client.embeddings.create(
        model=model,
        input=texts
    )
    return [data.embedding for data in response.data]

#EMBEDDING_MODEL = "text-embedding-3-small" 
```
