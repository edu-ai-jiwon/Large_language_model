# TTS (Text To Speech)

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from openai import OpenAI
client=OpenAI()
# 원하는 입력 텍스트 넣기 
input_text='봄동비빔밥의 핵심은 재료가 가진 달큰 꼬소한 신선한 맛을 최대한 살리는 것이다.'

with client.audio.speech.with_streaming_response.create(
    model='gpt-4o-mini-tts',
    voice='nova',#공식 홈페이지에 11개의 voice 모델명이 있음(참고)
    input=input_text
) as response:
    response.stream_to_file('tts_output.mp3')   #파일로 저장할 수 있도록함 -> 파일 생성
```

```python
from IPython.display import Audio   #쥬피터 노트북에서 음성 들을 수 있음

display(Audio('tts_output.mp3', autoplay=True))
```

# STT (Speech To Text)

```python
with open('tts_output.mp3','rb') as f:  #파일 열기  #read-bite:파일을 bite 형태로 읽어 오는 것
    transcriptions=client.audio.transcriptions.create(
        model='whisper-1',
        file=f
    ) 

    print(transcriptions)
```

```python
import speech_recognition as sr 

recognizer=sr.Recognizer()

while True:
    with sr.Microphone() as source:
        print('말씀하세요.')
        audio=recognizer.listen(source)
        text=recognizer.recognize_google(audio, language='ko-KR')
        print(text)

        if text == '종료':
            break
        # 돌리고 나서 마이크에 가까이 가서 말하면 알아서 인식하고 text 결과를 내어 줌
```
