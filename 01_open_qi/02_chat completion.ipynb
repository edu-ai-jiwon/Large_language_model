```python
stream_response=client.chat.completions.create(
    model='gpt-4.1-mini',
    messages=[
        {'role':'user', 'content':'stream 테스트를 위한 아주 긴 응답 메시지를 보내주세요.'}
    ],
    max_completion_tokens=2000,
    stream=True
)
```

```python
stream_response
```

```python
for chunk in stream_response:
    #print(chunk)
    if chunk.choices[0].delta.content is not None:  #content=text
        print(chunk.choices[0].delta.content, end='')
    #조건 걸기(발생할지 모를 예외에 대한 처리하기)

print('----------------응답 완료------------')
# 채팅을 거듭하면 거듭할수록 답변 토큰 길이 더 길어짐
# 이전 내용도 쌓여서 오기 때문임
```

### Token Counting
- 한번의 프롬프트 입출력 토큰과 서비스 호출 빈도를 고려해 서비스 제공 비용 산정
- 비용 = ((입력 토큰 수 * 단가) +(출력 토큰수* 단가))*월 서비스 호출수

```python
import tiktoken
gpt35=tiktoken.encoding_for_model('gpt-3.5')
print(gpt35)
print(len(gpt35.encode('오늘은 즐거운 월요입니다!')))


gpt4o=tiktoken.encoding_for_model('gpt-4o')
print(gpt4o)
print(len(gpt4o.encode('오늘은 즐거운 월요입니다!')))

gpt5=tiktoken.encoding_for_model('gpt-5')
print(gpt5)
print(len(gpt5.encode('오늘은 즐거운 월요입니다!')))
```

```python
model='gpt-5.2'
try:
    enc=tiktoken.encoding_for_model(model)

except KeyError as e: # 존재하지 않아서 KeyError 발생 시
    print(f'[Warning] {model}을 찾을 수 없습니다.')
    enc=tiktoken.get_encoding('o200k_base')
```

```python
input_text=''

response=client.chat.completions.create(
    model=model,
    messages=[
        {'role':'system', 'content':'당신은 시사경제 전무가로, 주어진 뉴스 기사의 핵심을 잘 요약/정리해주는 챗봇입니다. 사용자 입력을 잘 읽고, 핵심 내용 위주로 요약하세요.'},
        {'role':'user', 'content': input_text},
        {'role':'user', 'content': input_text},
    ],
    temperature=0.2,
    max_completion_tokens=4096,
    top_p=1
)
output_text=response.choices[0].message.content
```

```python
input_text={"식품의약품안전처(처장 오유경, 이하 식약처)와 지식재산처(처장 김용선, 이하 지재처), 관세청(청장 이명구)은 위조화장품 대응을 위한 범부처 합동 설명회를 누리꿈스퀘어(서울 마포구 소재)에서 지난 12일 개최했다고 밝혔다. 전 세계적으로 큰 인기를 끌고 있는 K-화장품의 지식재산권을 보호하고, 위조 화장품 유통에 따른 기업과 소비자 피해를 예방하기 위해서다.식약처, 지재처, 관세청은 부처별 전문성에 기반해 위조 화장품 단계별 대응 체계를 구축하고 기관별 정책 방향, 지원 사업을 안내했다.식약처는 유통 위조화장품의 사후관리 단계에서 추진 방향으로 위조화장품 판매자 처벌과 회수·폐기 조치 명령 등에 대한 명시적 법적 근거를 마련하고, 화장품 업계가 위조화장품 유통 사실을 확인한 경우 관련 내용을 제보할 수 있도록 대한화장품협회에 위조화장품 제보 센터를 설치·운영한다고 밝혔다.지식재산처는 화장품 분야 K-브랜드 침해 사례 및 상표·디자인권의 중요성 안내와 K-브랜드 침해 대응방안 및 지원사업·K-브랜드 보호포털을 소개하고, 해외 특허분쟁 동향 및 지식재산처 지원정책에 대해 소개하는 시간을 가졌다.관세청은 K-뷰티 기업의 지식재산권 보호를 위해 주요 수출국 관세당국에 상표 등 지식재산권을 등록할 수 있도록 안내했고, 해외 단속기관과의 글로벌 네트워크를 활용해 현지에서 K-브랜드 위조 화장품이 제조․유통되거나 수출입되지 않도록 합동 단속망을 구축할 계획이라고 밝혔다."}
```
