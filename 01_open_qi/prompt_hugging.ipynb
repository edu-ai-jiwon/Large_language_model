```python
from openai import OpenAI

#client=OpenAI(api_key=OPENAI_API_KEY)
client=OpenAI()
# 채팅 요청을 보내는 메서드
# create를 통해 객체 호출
response=client.chat.completions.create(
    # 파라미터
    model='gpt-4.1-mini',
    messages=[
        {
            'role':'system',
            'content':[{
                'type':'text',
                'text':'당신은 곧 멸망할 세상의 운명을 알고 있는 용사입니다. 세상의 끝이 온다는 걸 알지만 아무것도 할 수 없어 절망했다가 체념했습니다.'
            }]
        },
        {
            'role':'user',
            'content':[{
                'type': 'text',
                'text': '봄이 오는데 뭘 하면 좋을까?'
            }]
        },
    ],
    response_format={
        'type': 'text'
    },  #dic으로 전달
    temperature=1.0,                        #0~2 사이 값/ 기본값: 1 / 창의성
    top_p=1,
    max_tokens=2048,                        # 전체 토큰이 아님/응답 최대 토큰 수 지정
    frequency_penalty=0,                    # 토큰 재사용에 대한 패널티 -> 많이 사용할 수록 패널티 줌
    presence_penalty=0                      # 토큰 재사용에 대한 패널티 -> 한 번이라도 쓰면 패널티를 줌 = 같은 단어 사용 억제하게 해주는 파라미터
)
```

### 1. 기사 제목 교정

```python
from openai import OpenAI
#매번 같은 방식을 사용하기 위해 만든 것
client=OpenAI(api_key=OPENAI_API_KEY)
query=input('[기사 제목 교정 chatbot] 기사 내용을 넣어주세요! 내용을 토대로 기사 제목을 교정해줍니다!')   # 질문 입력
# input: 한 노인이 길을 가다가 차에 치일 위험이 있었다. 
# 다행히 근처에 있던 청년이 잽싸게 몸을 날려서 노인을 구했다. \
# 이런 일이 있었는데 뉴스로 보도할 때 기사 제목을 어떻게 하면 좋을까

# 문자열로 만들기
system_instruction='당신은 10년 베테랑 기자입니다. 여러 가지의 다양한 카테고리 기사 제목을 작성해야 합니다. ' \
'전해 받은 보도 자료를 토대로 수많은 뉴스 기사 독자가 관심을 가질만한 기사 제목을 작성해야 합니다.' \
'당신은 다양한 카테고리의 기사의 제목을 작성할 때, 따로 분리해서 정리하는 것을 좋아합니다.' \
'출퇴근 길 등 짧은 시간 내에 뉴스를 보는 독자들을 위한 기사 제목을 작성하세요.'   # 페르소나, 형식 지정
user_message=query # query에 덧붙이고 싶은 경우 수정함
#최종적으로 어떤 것을 수행하라는 질의

response=client.chat.completions.create(
    model='gpt-4.1-mini',
    messages=[
        {
            'role': 'system',
            'content': [{
                'type': 'text',
                'text':system_instruction
            }]
        },
        {
            'role':'user',
            'content':[{
                'type':'text',
                'text': user_message
            }]
        }
    ],
    response_format={
        'type': 'text'
    },  #dic으로 전달
    top_p=1,
    max_tokens=2048,
    frequency_penalty=0,
    presence_penalty=0,
)

print(response.choices[0].message.content)
```
